### SUDDIVISIONE PER ANNI ###

```python
import pandas as pd
shark_clean = pd.read_csv('shark_clean.csv')
shark_clean
```

```python
shark_clean.groupby("Year")['Fatal Y/N'].value_counts()
```

```python
fataYN = shark_clean.groupby("Sex")["Fatal Y/N"].value_counts()
fataYN
fataYN.unstack()

```

```python
fatal_percentage = (
    shark_clean
    .groupby("Sex")["Fatal Y/N"]
    .value_counts(normalize=True)
    .mul(100)
    .unstack(fill_value=0)
    .round(2)
)

fatal_percentage
```

```python
fatal_percentage.columns = ["No Fatal %", "Unknown", "Yes Fatal %" ]
fatal_percentage
```

```python
shark_clean["Fatal Y/N"].value_counts(normalize=True) * 100
```

```python
fatal_percentage = (
    shark_clean
    .groupby(["Sex", "Fatal Y/N"])
    .size()
    .unstack(fill_value=0)
    .div(len(shark_clean))
    .mul(100)
    .round(2)
)

fatal_percentage.columns = [
    "No Fatal %",
    "Unknown %",
    "Yes Fatal %"
]

fatal_percentage
```

```python
known = shark_clean[shark_clean["Fatal Y/N"] != "Unknown"]
decade = known["Year"] // 10 * 10

table = pd.crosstab(decade, known["Fatal Y/N"])
table["fatality_rate"] = (table["Y"] / (table["Y"] + table["N"]) * 100).round(1)
table
```

```python
known = shark_clean[
    (shark_clean["Fatal Y/N"] != "Unknown") &
    (shark_clean["Year"] >= 1990)
]

table = pd.crosstab(known["Year"], known["Fatal Y/N"])

table["fatality_rate"] = (
    table["Y"] / (table["Y"] + table["N"]) * 100
).round(1)

table
```

```python
shark_clean["Country"].value_counts()
```

```python
shark_clean["Country"].value_counts()[shark_clean["Country"].value_counts() >= 70]
```

```python
known = shark_clean[shark_clean["Fatal Y/N"] != "Unknown"]

counts = known["Country"].value_counts()
common = counts[counts >= 50].index
subset = known[known["Country"].isin(common)]

table = (
    pd.crosstab(
        subset["Country"],
        subset["Fatal Y/N"],
        normalize="index"
    )
    .mul(100)
    .round(2)
)

average = table.mean()

# Tabella 1: percentuale Y dal valore più alto al più basso
table_by_y = table.sort_values("Y", ascending=False)
table_by_y.loc["Average"] = average

# Tabella 2: percentuale N dal valore più alto al più basso
table_by_n = table.sort_values("N", ascending=False)
table_by_n.loc["Average"] = average

display(table_by_y)
display(table_by_n)
```

```python
known = shark_clean[shark_clean["Fatal Y/N"] != "Unknown"]

table = pd.crosstab(known["Type"], known["Fatal Y/N"], normalize="index") * 100
table["n"] = known["Type"].value_counts()
table.round(1)
```

```python
activity_type = (
    pd.crosstab(
        shark_clean["Activity"],
        shark_clean["Type"],
        normalize="index"
    )
    .mul(100)
)

activity_provoked = activity_type[["Provoked"]].copy()

activity_provoked["n"] = (
    shark_clean["Activity"].value_counts()
)

activity_provoked = (
    activity_provoked
    .round(1)
    .sort_values("Provoked", ascending=False)
)

activity_provoked
activity_provoked = activity_provoked[
    activity_provoked["n"] >= 10
]
activity_provoked
```

```python
activity_type = (
    pd.crosstab(
        shark_clean["Activity"],
        shark_clean["Type"],
        normalize="index"
    )
    .mul(100)
)

activity_comparison = activity_type[
    ["Provoked", "Unprovoked"]
].copy()

activity_comparison["n"] = (
    shark_clean["Activity"].value_counts()
)

activity_comparison = (
    activity_comparison
    .round(1)
    .sort_values("Provoked", ascending=False)
)

activity_comparison
```
